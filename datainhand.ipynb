# An Analysis of Instagram Advertising Data

## Project Overview

For my final project, I collected data from my Instagram account in JSON format. I selected the file advertisers_using_your_activity_or_information.json because it contains information about advertisers and how they were matched with my Instagram profile. I found advertising data that has lists of advertisers and the different ways my account was associated with them. This data will allow me to explore how advertisers utilize personal information and online interactions for targeted advertising.

```python
import json
import pandas as pd

with open("advertisers_using_your_activity_or_information.json", "r") as file:
    data = json.load(file)
```

```python
type(data)
```

```python
data.keys()
```

```python
for key, value in data.items():
    print("Category:", key)
    print("Data type:", type(value))
    print("Number of entries:", len(value))
    print()
```

```python
for key, value in data.items():
    print(key)
    print(value[:3] if isinstance(value, list) else type(value))
    print()
```

```python
summary = pd.DataFrame({
    "Category": list(data.keys()),
    "Number of Entries": [len(value) for value in data.values()]
})

summary
```
